```python
def add(a, b=10): #b has default value
    return a + b

print(add(5)) #15
print(add(5, 7)) #12
```

```python
def square(n):
    """ return n multiplied by itself """  #docstring
    return n * n

print(square(5))
```

```python
import csv

with open("price.csv") as p:
    rows = list(csv.DictReader(p))

print(rows)
print('total price is:')
print(sum(int(r["price"]) for r in rows))
```

```python
import numpy as np

a = np.array([1, 2, 3, 4, 5, 6])
print(a)
print(a.shape, a.dtype)
print(a * 2) #multiplication without loop, all at once
print('mean:', a.mean())
print('standard deviation:', a.std())
print('maximum:', a.max())
print('minimum:', a.min())
print('sum:', a.sum())

print(a[a>2]) #boolean masking, returns all elements greater than 2
print(np.zeros(3), np.arange(0, 10, 2), np.linspace(0, 1, 5)) #zeros, arrange, line space
print(a.reshape(3, 2))
```

```python
def average(numbers):
    Total = 0
    for n in numbers:
        Total += n
    return Total / len(numbers)
```
