```python
scores = [90, 85, 78, 92, 88]
scores.sort()
print(scores)
scores.remove(88)
print(scores)
scores.append(95)
print(scores)   
print(scores[0])  # Accessing the first element
print(len(scores))  # Getting the length of the list
print(scores[1:3])  # Slicing the list to get elements from index 1 to 2
```

```python
for score in scores:
    print(score)  # Printing each score in the list
```

```python
for s in scores:
    print(s * 2)  # Printing each score multiplied by 2
    
```

```python
for i in range(1, 6):
    if i == 3:
        continue  # Skip the rest of the loop when i is 3
    print(i)  # Printing numbers from 1 to 5, skipping 3
print("Loop ended")  # Indicating the end of the loop
for i in range(1, 6):
    if i == 3:
        break  # Exit the loop when i is 3
    print(i)  # Printing numbers from 1 to 5, stopping at 3
```

```python
i = 1
while i <= 5:
    if i <= 3:
        i += 1
        continue  # Skip the rest of the loop when i is less than or equal to 3
    print(i)  # Printing numbers from 5 to 5, skipping first 3
    i += 1
print("Loop ended")  # Indicating the end of the loop
i = 1
while i <= 3:
    if i == 2:
        break  # Exit the loop when i is 2
    print(i)  # Printing numbers from 1 to 3, stopping at 2
    i += 1
print("Loop ended")  # Indicating the end of the loop
```

```python
word = "Python"
print(word[0])  # Accessing the first character
print(word[1:4])  # Slicing the string to get characters from index 1 to 3
print(word.upper())  # Converting the string to uppercase
print("red,wine".split(","))  # Splitting the string into a list
print("      Hi!!      ".strip())  # Removing leading and trailing whitespace

name = "Raksha"
print(f"My name is {name}")  # Using f-string for string formatting
```

```python
person = { "name": "Raksha", "age": 20, "city": "New York" }
print(person.get("name"))  # Accessing the value associated with the key "name"
print(person.keys())  # Getting all the keys in the dictionary
print(person.values())  # Getting all the values in the dictionary
print(person.items())  # Getting all the key-value pairs in the dictionary
```

```python
# sets stores unique values, so duplicates are automatically removed
fruits = {"apple", "banana", "orange", "apple"}
print(fruits)  # Output: {"apple", "banana", "orange"}
```

```python
#list comprehension
squares = [number * number for number in range(1, 6)]
print(squares)  # Output: [1, 4, 9, 16, 25]
```

```python
# multidimensional list (list of lists)
grid = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print(grid[0][1])  # Accessing the element in the first row and second column
print("nested loops")

#nested loops
for row in grid:
    for element in row:
        print(element)  # Printing each element in the grid
```

```python
price = [200, 500, 800, 400, 250] 
total = sum(price)  # Calculating the sum of all prices
print(total)  # Printing the total price

for p in price:
    if p > 400:
        print(p)
```

```python
price = [200, 500, 800, 400, 250, 100] 
def discount(price):
    return price - (price * 0.1)

for p in price:
    print(discount(p))  # Calling the discount function with each price in the list
```
